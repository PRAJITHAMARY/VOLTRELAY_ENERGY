```python
# Memory-efficient data types for the large swap_events dataset

swap_dtypes = {
    "event_id": "string",
    "rider_id": "string",
    "station_id": "string",
    "event_type": "category",
    "attempt_seq": "int8",
    "queue_wait_sec": "float32",
    "battery_in_id": "string",
    "battery_out_id": "string",
    "soc_in_pct": "float32",
    "soh_in_pct": "float32",
    "soc_out_pct": "float32",
    "soh_out_pct": "float32",
    "km_since_last_swap": "float32",
    "tariff_code": "category",
    "list_price_inr": "float32",
    "discount_inr": "float32",
    "amount_charged_inr": "float32",
    "payment_mode": "category",
    "energy_to_recharge_kwh": "float32",
    "station_firmware": "category",
    "sync_mode": "category"
}

swap = pd.read_csv(
    "/content/swap_events.csv",
    dtype=swap_dtypes,
    parse_dates=["event_ts"]
)

print("Swap dataset loaded successfully!")
print("Shape:", swap.shape)
print("Memory usage:", round(swap.memory_usage(deep=True).sum() / 1024**3, 2), "GB")
```

```python
print("Rows:", len(swap))
print("Columns:", len(swap.columns))

print("\nDate range:")
print(swap["event_ts"].min())
print(swap["event_ts"].max())

print("\nEvent types:")
display(swap["event_type"].value_counts())

print("\nTariff types:")
display(swap["tariff_code"].value_counts())

print("\nSync modes:")
display(swap["sync_mode"].value_counts())
```

```python
missing = swap.isna().sum().sort_values(ascending=False)

missing_df = pd.DataFrame({
    "missing_count": missing,
    "missing_percentage": (missing / len(swap) * 100).round(2)
})

display(missing_df)
```

```python
print("Exact duplicate rows:", swap.duplicated().sum())
print("Duplicate event IDs:", swap["event_id"].duplicated().sum())
```

```python
swap_clean = swap.copy()

print("Original rows:", len(swap_clean))
```

```python
swap_clean["is_test_station"] = swap_clean["station_id"].str.startswith("STN-TST")

print("Test-station events:", swap_clean["is_test_station"].sum())
```

```python
swap_clean["event_ts_corrected"] = swap_clean["event_ts"]

bug_mask = (
    (swap_clean["station_firmware"] == "v3.2.0") &
    (swap_clean["event_ts"] >= "2025-03-10") &
    (swap_clean["event_ts"] < "2025-04-15")
)

swap_clean.loc[bug_mask, "event_ts_corrected"] = (
    swap_clean.loc[bug_mask, "event_ts"] + pd.Timedelta(hours=5, minutes=30)
)

print("Timestamp-corrected records:", bug_mask.sum())
```

```python
invalid_distance = (
    (swap_clean["km_since_last_swap"] < 0) |
    (swap_clean["km_since_last_swap"] > 1000)
)

swap_clean["invalid_distance"] = invalid_distance

print("Invalid distance records:", invalid_distance.sum())
```

```python
soc_anomaly = (
    (swap_clean["soc_in_pct"] < 0) |
    (swap_clean["soc_in_pct"] > 100) |
    (swap_clean["soc_out_pct"] < 0) |
    (swap_clean["soc_out_pct"] > 100)
)

soh_anomaly = (
    (swap_clean["soh_in_pct"] < 0) |
    (swap_clean["soh_in_pct"] > 100) |
    (swap_clean["soh_out_pct"] < 0) |
    (swap_clean["soh_out_pct"] > 100)
)

print("SOC anomalies:", soc_anomaly.sum())
print("SOH anomalies:", soh_anomaly.sum())
```

```python
offline = swap_clean[
    swap_clean["sync_mode"] == "offline_batch"
].copy()

print("Offline-batch records:", len(offline))
```

```python
offline_sorted = offline.sort_values(
    ["rider_id", "station_id", "event_ts_corrected"]
)

offline_sorted["time_diff_sec"] = (
    offline_sorted
    .groupby(["rider_id", "station_id"])["event_ts_corrected"]
    .diff()
    .dt.total_seconds()
)

near_duplicates = offline_sorted[
    offline_sorted["time_diff_sec"].between(0, 180)
]

print("Potential offline near-duplicates:", len(near_duplicates))
```

```python
swap_clean["date"] = swap_clean["event_ts_corrected"].dt.date
swap_clean["month"] = swap_clean["event_ts_corrected"].dt.to_period("M").astype(str)
swap_clean["hour"] = swap_clean["event_ts_corrected"].dt.hour
swap_clean["day_of_week"] = swap_clean["event_ts_corrected"].dt.day_name()

print(swap_clean[[
    "event_ts_corrected",
    "date",
    "month",
    "hour",
    "day_of_week"
]].head())
```

```python
station_info = stations[[
    "station_id",
    "city",
    "zone",
    "location_type",
    "host_type",
    "expansion_wave",
    "charger_generation",
    "slots_2w",
    "slots_3w",
    "inventory_target_2w",
    "inventory_target_3w",
    "monthly_rent_inr",
    "monthly_maintenance_inr",
    "grid_tariff_inr_kwh",
    "connectivity_tier"
]].copy()

swap_clean = swap_clean.merge(
    station_info,
    on="station_id",
    how="left",
    validate="many_to_one"
)

print("Rows after station join:", len(swap_clean))
print("Missing station matches:", swap_clean["city"].isna().sum())
```

```python
network = swap_clean[
    ~swap_clean["is_test_station"]
].copy()

print("Network events:", len(network))
```

```python
network["is_completed"] = (
    network["event_type"] == "swap_completed"
).astype(int)

network["is_failed"] = (
    network["event_type"] != "swap_completed"
).astype(int)
```

```python
monthly_kpi = (
    network
    .groupby("month")
    .agg(
        total_attempts=("event_id", "count"),
        completed_swaps=("is_completed", "sum"),
        revenue_inr=("amount_charged_inr", "sum")
    )
    .reset_index()
)

monthly_kpi["completion_rate_pct"] = (
    monthly_kpi["completed_swaps"]
    / monthly_kpi["total_attempts"]
    * 100
)

monthly_kpi["failure_rate_pct"] = (
    100 - monthly_kpi["completion_rate_pct"]
)

monthly_kpi["revenue_per_completed_swap"] = (
    monthly_kpi["revenue_inr"]
    / monthly_kpi["completed_swaps"]
)

display(monthly_kpi)
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_kpi["month"],
    monthly_kpi["completed_swaps"],
    marker="o"
)

plt.title("Monthly Completed Swaps")
plt.xlabel("Month")
plt.ylabel("Completed Swaps")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_kpi["month"],
    monthly_kpi["failure_rate_pct"],
    marker="o"
)

plt.title("Monthly Swap Failure Rate")
plt.xlabel("Month")
plt.ylabel("Failure Rate (%)")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_kpi["month"],
    monthly_kpi["revenue_inr"] / 1e6,
    marker="o"
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹ Million)")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
failure_monthly = (
    network
    .groupby(["month", "event_type"])
    .size()
    .reset_index(name="count")
)

display(failure_monthly)
```

```python
failure_monthly_pct = (
    failure_monthly
    .pivot(index="month", columns="event_type", values="count")
    .fillna(0)
)

failure_monthly_pct["total_attempts"] = (
    network.groupby("month").size()
)

for col in [
    "failed_no_charged_battery",
    "abandoned_queue",
    "cancelled_by_rider",
    "failed_system_error"
]:
    if col in failure_monthly_pct.columns:
        failure_monthly_pct[col + "_pct"] = (
            failure_monthly_pct[col]
            / failure_monthly_pct["total_attempts"]
            * 100
        )

display(failure_monthly_pct)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)


categorical_features = [
    "vehicle_class",
    "day_of_week",
    "city",
    "tariff_code"
]

numeric_features = [
    "hour",
    "queue_wait_sec",
    "soc_in_pct",
    "soh_in_pct",
    "km_since_last_swap"
]


preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", "passthrough", numeric_features)
    ]
)


X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Encoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

model.fit(X_train_encoded, y_train)

y_pred = model.predict(X_test_encoded)
y_prob = model.predict_proba(X_test_encoded)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

## Business Recommendations

### 1. Reduce Swap Failures Through Queue Monitoring
Queue wait time is one of the strongest predictors of swap failure.
VoltRelay should monitor stations in real time and trigger alerts when
queue conditions become abnormal.

### 2. Improve Battery Availability
The analysis shows that battery availability-related failures are a
major contributor to failed swaps, particularly under extreme queue
conditions. Stations should maintain sufficient charged-battery
inventory during high-demand periods.

### 3. Prioritize Low-SOH Batteries
Battery health should be continuously monitored. Batteries with
significant SOH degradation should be inspected, rotated out of
high-demand stations, or retired before they affect service reliability.

### 4. Investigate Supplier and Manufacturing-Lot Risk
Kyron batteries show substantially higher SOH degradation compared
with Cellora and Amptek. VoltRelay should investigate Kyron battery
lots, manufacturing quality, and lifecycle performance.

### 5. Prioritize High-Risk Stations
Stations with unusually high failure rates should be investigated for
battery availability, equipment problems, demand spikes, and operational
issues.

### 6. Improve 3W Service Reliability
3W swaps show significantly higher failure rates during extreme queue
conditions. Additional battery inventory and capacity should be
considered for high-demand 3W locations.

### 7. Deploy the ML Risk Prediction Model
The ML model can identify high-risk swap attempts before failure.
Using the optimized probability threshold of 0.30 improves the balance
between precision and recall, making it useful for proactive alerts.

### 8. Build a Proactive Failure Management System
Combine queue conditions, battery SOH, vehicle class, station,
city, and other operational features into a real-time risk score.
This can help operations teams intervene before a swap fails.

## Conclusion

The analysis shows that swap failures are driven by a combination of
queue conditions, battery health, station-level operations, and vehicle
characteristics.

Queue conditions are particularly important, with both extremely short
and very long queues associated with significantly higher failure rates.
Battery degradation also creates operational risk, with clear differences
between suppliers and pack types.

The machine learning model achieved a ROC-AUC of approximately 0.969,
showing strong ability to distinguish high-risk swap attempts. After
threshold optimization, a threshold of 0.30 provided a better balance
between precision and recall.

Overall, VoltRelay can use these insights to move from reactive failure
handling toward proactive intervention, improving swap reliability,
rider experience, and operational efficiency.
